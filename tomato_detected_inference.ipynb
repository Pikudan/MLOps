```python
import ultralytics
from ultralytics import RTDETR, YOLO
```

```python
def inference_RTDETR():
    model = RTDETR("C:/Users/Admin/AgrlinkCabbageYOLOe/rtdetr_l3/weights/best.pt")
    metrics = model.val(
        # Project
        project="cabbagetestrtdetr",
        name="cabbagetestrtdetr",
        task='detect',
        # Random Seed parameters
        # Data & model parameters
        data='C:/Users/Admin/Corn-17/data.yaml',
        split='test',
        save_json=True,
        save=True,
        imgsz=416,
        batch=16,
        iou=0.2,
        device='0',
        augment=False,
        workers = 4,
    )
    return metrics
```

```python
def inference_cabbage(iou):
    model = YOLO("C:/Users/Admin/AgrlinkCabbageYOLOe/yolov8nv1/weights/best.pt")
    metrics = model.val(
        # Project
        project="cabbagetestrtdetr",
        name="cabbagetestrtdetr",
        task='detect',
        # Random Seed parameters
        # Data & model parameters
        data='C:/Users/Admin/Corn-17/data.yaml',
        split='test',
        save_json=True,
        save=True,
        imgsz=416,
        batch=16,
        iou=iou, #for nms
        device='0',
        augment=False,
        workers = 4,
    )
    return metrics
```

```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"]="TRUE"
```

```python
import numpy as np
metrics_from_iou = []
for iou in np.linspace(0.1, 1, 10):
    metrics_from_iou.append(inference_cabbage(iou))
```

```python
mAP = [metrics.box.all_ap for metrics in metrics_from_iou]
```

```python
mAP = np.array(mAP)
```

```python
np.save("map_from_iou", mAP)
```
